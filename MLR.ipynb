```python
import pandas as pd
```

```python
dataset=pd.read_csv("50_Startups_dataset.csv")
```

```python
dataset
```

```python
dataset=dataset.drop(columns=['Unnamed: 0'])
```

```python
dataset
```

```python
dataset.columns
```

```python
dataset=pd.get_dummies(dataset,drop_first=True)
```

```python
dataset
```

```python
dataset.columns
```

```python
independent=dataset[['R&D Spend', 'Administration', 'Marketing Spend','State_Florida', 'State_New York']]
dependent=dataset[['Profit']]
```

```python
independent
```

```python
dependent
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(independent,dependent,test_size=0.2,random_state=11)
```

```python
x_train
```

```python
y_train
```

```python
x_test
```

```python
y_test
```

```python
from sklearn.linear_model import LinearRegression
```

```python
regressor=LinearRegression() 
regressor.fit(x_train,y_train)
```

```python
y_pred=regressor.predict(x_test)
y_pred
```

```python
from sklearn.metrics import r2_score
r2score=r2_score(y_test,y_pred)
```

```python
r2score
```

```python
import pickle
filename="MLR.sav"
pickle.dump(regressor,open(filename,'wb'))
```

```python
loaded_model=pickle.load(open("MLR.sav",'rb'))
result=loaded_model.predict([[1000,2000,3000,4000,5000]])
```
